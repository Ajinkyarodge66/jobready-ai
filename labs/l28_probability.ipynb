```python
import numpy as np
rng = np.random.default_rng(42)
```

## 1. Bayes' theorem, by simulation
Disease prevalence 1%, sensitivity 90%, false-positive rate 5%.

```python
n = 1_000_000
sick = rng.random(n) < 0.01
positive = np.where(sick, rng.random(n) < 0.90, rng.random(n) < 0.05)
print('P(sick | positive) ≈', sick[positive].mean())   # about 0.154
print('exact', 0.9*0.01 / (0.9*0.01 + 0.05*0.99))
```

## 2. Distributions

```python
print('binomial mean', rng.binomial(20, 0.3, 100_000).mean(), '(expect 6)')
print('poisson mean/var', rng.poisson(4, 100_000).mean(), rng.poisson(4, 100_000).var())
z = rng.normal(0, 1, 100_000)
print('within 1, 2, 3 sd', [(abs(z) < k).mean().round(3) for k in (1, 2, 3)])
```

## 3. Maximum likelihood for a coin
Which p makes 7 heads in 10 flips most likely?

```python
from math import comb
ps = np.linspace(0.01, 0.99, 99)
lik = [comb(10, 7) * p**7 * (1 - p)**3 for p in ps]
print('MLE p =', ps[int(np.argmax(lik))])   # 0.7
```
