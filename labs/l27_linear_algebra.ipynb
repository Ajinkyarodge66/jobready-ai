```python
import numpy as np
np.set_printoptions(precision=3, suppress=True)
```

## 1. Vectors

```python
x = np.array([3, 4]); y = np.array([4, 3])
print('dot', x @ y)
print('norms', np.linalg.norm(x), np.linalg.norm(y))
print('cosine', x @ y / (np.linalg.norm(x) * np.linalg.norm(y)))
```

## 2. Matrices as transformations
The columns of `A` show where the basis vectors land.

```python
A = np.array([[2, 0], [1, 1]])
square = np.array([[0, 1, 1, 0], [0, 0, 1, 1]])   # corners of the unit square as columns
print(A @ square)
print('determinant (area scale):', np.linalg.det(A))
```

## 3. Shapes
Read each shape before you multiply.

```python
X = np.random.randn(100, 5)      # 100 rows, 5 features
w = np.random.randn(5)
print((X @ w).shape)              # (100,)
W = np.random.randn(5, 3)
print((X @ W).shape)              # (100, 3)
```

## 4. PCA from scratch

```python
rng = np.random.default_rng(0)
X = rng.normal(size=(500, 2)) @ np.array([[3, 1], [1, 0.5]])
Xc = X - X.mean(axis=0)
cov = Xc.T @ Xc / (len(Xc) - 1)
vals, vecs = np.linalg.eigh(cov)
order = vals.argsort()[::-1]
vals, vecs = vals[order], vecs[:, order]
print('explained variance ratio', vals / vals.sum())
Z = Xc @ vecs[:, :1]   # project onto the first component
print(Z.shape)
```
