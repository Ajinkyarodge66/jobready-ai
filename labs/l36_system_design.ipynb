## Choose a threshold from business costs
A missed fraud costs £120; a wrongly declined genuine payment costs £8 (lost sale and an unhappy customer).

```python
import numpy as np
rng = np.random.default_rng(1)
n = 200_000; fraud = rng.random(n) < 0.004
score = np.where(fraud, rng.beta(5, 2, n), rng.beta(1.5, 8, n))
best = None
for t in np.linspace(0.05, 0.95, 91):
    flag = score >= t
    cost = 120 * (fraud & ~flag).sum() + 8 * (~fraud & flag).sum()
    if best is None or cost < best[1]: best = (t, cost)
t = best[0]; flag = score >= t
print(f'best threshold {t:.2f}, recall {(flag & fraud).sum() / fraud.sum():.2%}, false-positive rate {(flag & ~fraud).sum() / (~fraud).sum():.2%}')
```
